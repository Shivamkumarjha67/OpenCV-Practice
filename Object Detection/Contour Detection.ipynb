# CONTOUR DETECTION -> OUTLINES

### Allows us to detect foreground vs background images

### Contours -> A Curve joining all the continuous points (along the boundary), having same color or intensity.

### Useful for shape analysis, object detection and recognition.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
img = cv2.imread('../DATA/internal_external.png', 0)
plt.imshow(img, cmap='gray')
```

```python
contours, hierarchy = cv2.findContours(img, cv2.RETR_CCOMP, cv2.CHAIN_APPROX_SIMPLE)
print(type(contours))
print(type(hierarchy))
print(len(hierarchy))
```

```python
len(contours)
```

```python
# In hierarchy, we can see the grouping of contours by last value of each hierachy (i.e, -1, 4, 0)
hierarchy
```

```python
# plt.imshow(img)
```

```python
external_contours = np.zeros(img.shape)
external_contours.shape
```

```python
for i in range(len(contours)):
    # External contours
    if hierarchy[0][i][3] == -1:
        cv2.drawContours(external_contours, contours, i, 255, -1)
```

```python
# Drawing the external contours foound
plt.imshow(external_contours, cmap='gray')
```

```python
internal_contours = np.zeros(img.shape)

for i in range(len(contours)):
    # Internal contours
    if hierarchy[0][i][3] != -1:
        cv2.drawContours(internal_contours, contours, i, 255, -1)
```

```python
# Showing the internal contours
plt.imshow(internal_contours, cmap='gray')
```
