# GRID DETECTION

### Detecting corners and edges both concepts to find grid in images

### Grid patterns are often use to caliberate cameras and track motion. 

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
flat_chess = cv2.imread('../DATA/flat_chessboard.png')
plt.imshow(flat_chess)
```

```python
 # Finding corners of the chessboard
found, corners = cv2.findChessboardCorners(flat_chess, (7,7))
```

```python
# Drawing corners, that are found
cv2.drawChessboardCorners(flat_chess, (7,7), corners, found)
plt.imshow(flat_chess)
```

```python
dots = cv2.imread('../DATA/dot_grid.png')
plt.imshow(dots)
```

```python
# Find Corners
found, corners = cv2.findCirclesGrid(dots, (10, 10), cv2.CALIB_CB_SYMMETRIC_GRID)
# Checking found or not
found
```

```python
cv2.drawChessboardCorners(dots, (10, 10), corners, found)
plt.imshow(dots)
```
