# EDGE DETECTION

### Expanding to find the general edges of objects

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
img = cv2.imread('../DATA/sammy_face.jpg')
plt.imshow(img)
```

```python
# Canny Edge Detection -> General Approach, with lots of edge noise from env as well.
edges = cv2.Canny(image=img, threshold1=127, threshold2=127)
plt.imshow(edges)
```

```python
# Check, manipulating the threshold values
edges = cv2.Canny(image=img, threshold1=0, threshold2=255)
plt.imshow(edges)
```

```python
# Getting the median value, for the different color intensity.
med_val = np.median(img)
med_val
```

```python
# Lower THRESHOLD to either 0 or 70% of the MEDIAN Value whichever is greater.
lower = int(max(0, 0.7*med_val))
# Upper THRESHOLD to either 130% of the median or the max 255, whichever is smaller.
upper = int(min(255, 1.3*med_val))
```

```python
edges = cv2.Canny(image=img, threshold1=lower, threshold2=upper)
plt.imshow(edges)
```

```python
# Not performing great, First try blurring the original image.
blurred_img = cv2.blur(img, ksize=(5,5)) # Manipulate blurring, by changing kernel size.
```

```python
# Performs better than previous one, try manipulating the threshold values to get enhanced results.
edges = cv2.Canny(image=blurred_img, threshold1=lower, threshold2=upper+50)
plt.imshow(edges)
```
