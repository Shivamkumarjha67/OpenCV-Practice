# DRAWING CIRCLE ON THE LIVE FEED VIDEO

```python
import cv2

# Defining the mouse callback function
def draw_circle(event, x, y, flags, param):
    global clicked, centre
    
    if event == cv2.EVENT_LBUTTONDOWN:
        clicked = False
        centre = (x,y)
    elif event == cv2.EVENT_LBUTTONUP:
        clicked = True        

# Defining the centre and boolean clicked variable, haven't clicked anything yet!
centre = (0,0)
clicked = False

# CATCHING THE LIVE VIDEO
cap = cv2.VideoCapture(0)

cv2.namedWindow('Practice')
cv2.setMouseCallback('Practice', draw_circle)

while True:
    ret, frame = cap.read()

    if clicked:
        cv2.circle(frame, centre, radius=50, color = (0, 0, 255), thickness=4)

    cv2.imshow('Practice', frame)

    if cv2.waitKey(1) == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
