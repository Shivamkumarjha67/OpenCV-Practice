```python
import cv2

cap = cv2.VideoCapture(0)

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Creating the writer object, for storing the video
writer = cv2.VideoWriter('myrecordedvideo.mp4', cv2.VideoWriter_fourcc(*'DIVX'), 20, (width, height))

while True:
    ret, frame = cap.read()
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Operation of saving frame by frame in writer object
    writer.write(frame)

    # in Gray coloring
    # cv2.imshow('frame', gray)

    # in normal coloring
    cv2.imshow('frame', frame)

    # Killing window using escape button
    if cv2.waitKey(1) == 27:
        break

cap.release()
cv2.destroyAllWindows()
```

```python
# Playing the video that was recorded previously.
import cv2
import time

cap = cv2.VideoCapture("C:\\Users\\shivam kumar jha\\OneDrive\\Desktop\\CV Practice\\Video Basics\\myrecordedvideo.mp4")

if cap.isOpened() == False:
    print('ERROR FILE NOT FOUND OR WRONG CODEC USED!')

while cap.isOpened():
    ret, frame = cap.read()

    if ret == True:
        # Delaying thr video processing, since videoCapture is generally used for the video processing, by reading each frame.
        time.sleep(1/20)
        cv2.imshow('frame', frame)
        # Killing window using escape button
        if cv2.waitKey(1) == 27:
            break
    else:
        break

cap.release()
cv2.destroyAllWindows()
```
