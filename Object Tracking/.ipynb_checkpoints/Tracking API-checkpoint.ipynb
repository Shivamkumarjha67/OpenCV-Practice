# TRACKING API

```python
import cv2
```

```python
# Tracker selection
def ask_for_tracker():
    print('Select tracker:')
    print('0: MIL')
    print('1: KCF')
    print('2: CSRT')

    choice = input('Enter choice: ')

    if choice == '0':
        tracker = cv2.TrackerMIL_create()
        name = "TrackerMIL"

    elif choice == '1':
        tracker = cv2.TrackerKCF_create()
        name = "TrackerKCF"

    elif choice == '2':
        tracker = cv2.TrackerCSRT_create()
        name = "TrackerCSRT"

    else:
        tracker = cv2.TrackerKCF_create()
        name = "TrackerKCF (default)"

    return tracker, name
```

```python
tracker, name = ask_for_tracker()
```

```python
name.split()[0]
```

```python
tracker_name = name.split()[0]

cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Cannot open camera")
    exit()

ret, frame = cap.read()

if not ret:
    print("Failed to read first frame")
    exit()

roi = cv2.selectROI("Select ROI", frame, False)

if roi == (0, 0, 0, 0):
    print("ROI not selected")
    exit()

tracker.init(frame, roi)

while True:
    ret, frame = cap.read()

    if not ret:
        break

    success, roi = tracker.update(frame)

    (x, y, w, h) = tuple(map(int, roi))

    if success:
        cv2.rectangle(frame, (x,y), (x+w, y+h), (0,255,0), 3)
    else:
        cv2.putText(frame, "Tracking Failure!", (100,200),
                    cv2.FONT_HERSHEY_SIMPLEX, 1, (0,0,255), 3)

    cv2.putText(frame, tracker_name, (20,40),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0,255,0), 2)

    cv2.imshow("Tracking", frame)

    if cv2.waitKey(10) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
