# IMAGE THRESHOLDING 

### Thresholding -> Segmenting an image into different parts

### It converts an image to consists of only two color, white and black (means binary).

```python
import cv2
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
img = cv2.imread('../DATA/rainbow.jpg')
plt.imshow(img)
```

```python
img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(img)
```

```python
plt.imshow(img, cmap='gray')
```

```python
ret, thresh1 = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)
```

```python
ret
```

```python
plt.imshow(thresh1, cmap='gray')
```

```python
ret, thresh2 = cv2.threshold(img, 127, 255, cv2.THRESH_TRUNC)
```

```python
plt.imshow(thresh2, cmap='gray')
```

```python
ret, thresh3 = cv2.threshold(img, 127, 255, cv2.THRESH_TOZERO)
```

```python
plt.imshow(thresh3, cmap='gray')
```

```python
ret, thresh4 = cv2.threshold(img, 127, 255, cv2.THRESH_TOZERO_INV)
```

```python
plt.imshow(thresh4, cmap='gray')
```

```python
img = cv2.imread('../DATA/crossword.jpg', 0)
plt.imshow(img, cmap='gray')
```

```python
def show_pic(img):
    fig = plt.figure(figsize=(15, 15))
    ax = fig.add_subplot(111)
    ax.imshow(img, cmap='gray')
```

```python
show_pic(img)
```

```python
ret, th1 = cv2.threshold(img, 180, 255, cv2.THRESH_BINARY)
show_pic(th1)
```

```python
th2 = cv2.adaptiveThreshold(img, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 11, 8)
show_pic(th2)
```

```python
blended = cv2.addWeighted(src1=th1, alpha=0.6, src2=th2, beta=0.4, gamma=0)
show_pic(blended)
```
