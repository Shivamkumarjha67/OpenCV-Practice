```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
def display_img(img, cmap=None):
    fig = plt.figure(figsize=(12,10))
    ax = fig.add_subplot(111)
    ax.imshow(img, cmap)
```

```python
giraffe = cv2.imread('../DATA/giraffes.jpg')
```

```python
# Displaying giraffe image in RGB format
giraffe = cv2.cvtColor(giraffe, cv2.COLOR_BGR2RGB)
display_img(giraffe)
```

```python
# Applying the binary threshold
giraffe_gray = cv2.cvtColor(giraffe, cv2.COLOR_BGR2GRAY)
ret, thresh = cv2.threshold(giraffe_gray, 127, 255, cv2.THRESH_BINARY)
display_img(thresh, 'gray')
```

```python
hsv_giraffe = cv2.cvtColor(giraffe, cv2.COLOR_BGR2HSV)
```

```python
display_img(hsv_giraffe, 'gray')
```

```python
giraffe = cv2.imread('../DATA/giraffes.jpg')
```

```python
color = ['b', 'g', 'r']

for i, col in enumerate(color):
    hist = cv2.calcHist([giraffe], [i], None, [256], [0, 256])
    plt.plot(hist, color = col)
    plt.xlim([0, 256])

plt.title('HISTOGRAM FOR GIRAFFE')
```
